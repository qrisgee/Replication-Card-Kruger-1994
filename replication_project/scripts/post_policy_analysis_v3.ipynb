```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
```

```python
state_by_month_cps = pd.read_csv('/Users/chrisgee/UCSD2526/replication_project/data/state_by_month_cps.csv')
state_by_month_cps = state_by_month_cps[state_by_month_cps['IND1990'] == 641]
state_by_month_cps = state_by_month_cps[['PERIOD', 'POLICY_PERIOD', 'NJ_PA', 'EMPSTAT', 'DID_DUMMY', 'RECESSION_PERIOD', 'WTFINL']]
state_by_month_cps['PERIOD'] = pd.to_datetime(state_by_month_cps['PERIOD']).dt.to_period('M')
state_by_month_cps
```

```python
# Setup for the DiD estimation. In this version of the estimation I adjusted the estimation to include heterskedastic consistent estimators
# Included a dummy variable to account for the recession that ocurred between 1990 - 1991 to avoid any instance of an omitted variable bias

did_formula = 'EMPSTAT ~ NJ_PA + POLICY_PERIOD + DID_DUMMY + RECESSION_PERIOD'
model = smf.wls(did_formula, data=state_by_month_cps, weights=state_by_month_cps['WTFINL']).fit(cov_type='HC1')

with open('POST_POLICY_DID_ANALYSIS', 'w') as f:
    f.write(model.summary().as_text())
```
